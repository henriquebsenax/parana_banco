● **Passo 3:** Neste passo, estamos suspeitando de que esta base tem como chave primária a variável "contrato". Desta forma, vamos fazer uma verificação da unicidade desta variável. Para isso, vamos verificar se a combinação da variável "contrato" com alguma outra variável gera alguma duplicação de valores.

```python
### VERIFICAÇÃO DE UNICIDADE DA VARIÁVEL "contrato"
qtd_aux_list = []
for col in [col for col in DF_ORIGINAL_DATA.columns if col!='contrato']:
    DF_CONT_AUX = DF_ORIGINAL_DATA[['contrato', col]].drop_duplicates().groupby(by='contrato')[col].count().sort_values(ascending=False).reset_index()
    DF_CONT_AUX.columns = ['contrato', 'qtd_'+col]
    DF_CONT_AUX = DF_CONT_AUX[DF_CONT_AUX['qtd_'+col]>1]
    qtd_cont_aux = DF_CONT_AUX.shape[0]
    qtd_aux_list.append(qtd_cont_aux)
    if qtd_cont_aux != 0:
        print('Contrato vs', col)
        print('')

if np.sum(qtd_aux_list) == 0:
    print('A combinação da vairável "Contrato" e qualquer outra coluna da base de dados, sempre gera um registro único.')
```

**Conclusão:** Vemos que a variável contrato pode ser entendida como chave primária desta base de dados e condiz com o que é esperado na tarefa 3. Desta forma, guiaremos todo o estudo entendendo que o contrato é a observação e as demais variáveis são características do contrato. Além disso, iremos entender que neste contexto, um contrato é uma operação.

```python
### FUNÇÃO PARA CALCULAR DISTRIBUIÇÃO DE FREQUÊNCIA PARA UMA VARIÁVEL QUANTITATIVA
#import pandas as pd
#import numpy as np
#import math 

def dist_freq_h_fun(x_, var_int = False, type_ = 'inequality', k_ = 5, col_names = True, col_names_vet = ['Abs_Freq', 'Percent_Freq'], n_cols_ = 2):

    # PARÂMETRO DE ENTRADA DE DADOS.
    ## x_: Um objeto do tipo numpy.array ou pandas.serie.

    # PARÂMETRO DE CONTROLE DO TIPO DE VARIÁVEL.
    ## var_int: Este parâmetro define se você está trabalhando com uma variável inteira. Por padrão temos var_int= False, assumindo que a variável é contínua.

    # PARÂMETROS PARA A ESTIMAÇÃO DO NÚMERO DE CLASSES NA DISTRIBUIÇÃO DE FREQUÊNCIA.
    ## type_: String com o nome da técnica para estimar a quantidade de categorias a ser utilizada na Distribuição de Frequência. 
        ### Por padrão temos type_ = "inequality", mas podemos usar:
            #### inequality - Critério da desigualdade para calcular o número de classes na distribuição;
            #### subjective - Critério subjetivo. Aqui o usuário define o número de classes na distribuição; 
            #### sturges - Regra de Sturges para calcular o número de classes na distribuição;
            #### root_n - Critério da raiz quadrada. Aqui o número de classes na distribuição será o menor número interio que seja maior que a raiz quadrada do número de observações.

    ## k_: Número de classes para a distribuição de frequência. Por padrão temos k_ = 5. Este parâmetro só funciona se type_="subjective".


    # PARÂMETROS DE CONTROLE DO RESULTADO: Aqui são parâmetros que controlam o que vai sair no dataframe da distribuição de frequência
    ## col_names: Este parâmetro informa se você vai querer definir quais colunas você vai querer no dataframe da distribuição de frequência. Por padrão temos col_names = True.

    ## col_names_vet: Este deve ser uma lista de strings informando as colunas que você vai querer no dataframe da distribuição de frequência. 
        ### Por padrão temos col_names_vet = ['Abs_Freq', 'Percent_Freq'], mas os nomes possíveis são:
            #### 'Abs_Freq': Frequência Absoluta
            #### 'Cum_Abs_Freq': Frequência Absoluta Acumulada
            #### 'Relat_Freq': Frequência Relativa
            #### 'Cum_Relat_Freq': Frequência Relativa Acumulada
            #### 'Percent_Freq': Frequência Percentual
            #### 'Cum_Percent_Freq': Frequência Percentual Acumulada

    ## n_cols_: Este parâmetro remete ao númer de colunas que você vai querer ver no detaframe da distribuição de frequência. Este parâmetro só funciona se col_names = False.
        ###Por padrão temos n_cols_ = 2, mas pode ser um número inteiro de 1 a 5. Para estes números, você terá o seguinte resultado no dataframe:
            #### 1 - 'Abs_Freq'
            #### 2 - 'Abs_Freq', 'Cum_Abs_Freq'
            #### 3 - 'Abs_Freq', 'Cum_Abs_Freq', 'Relat_Freq'
            #### 4 - 'Abs_Freq', 'Cum_Abs_Freq', 'Relat_Freq', 'Cum_Relat_Freq'
            #### 5 - 'Abs_Freq', 'Cum_Abs_Freq', 'Relat_Freq', 'Cum_Relat_Freq', 'Percent_Freq'
            #### Qualquer outro número - 'Abs_Freq', 'Cum_Abs_Freq', 'Relat_Freq', 'Cum_Relat_Freq', 'Percent_Freq', 'Cum_Percent_Freq'

    
    # CÓDIGOS DA FUNÇÃO
    if type(x_) == pd.Series:
        x_ = x_.values
    elif type(x_) != np.ndarray:
        return 'Entre com uma variável do tipo np.array ou pd.serie.'

    if pd.DataFrame(x_, columns=['x_'])['x_'].dtypes == 'O':
        return 'Coloque um vetor (array) com valores numericos.'
    
    x_ = x_[~np.isnan(x_)]

    if len(x_)<1:
        return 'Sem observacoes'

    if type_ == 'subjective':
        c_ = k_

    elif type_ == 'sturges':
        try:         
            c_ = 3.3 * np.log(len(x_))
            if c_%1 != 0:
                c_ = int(c_) + 1

        except:
            return 'Problema com o logaritmo do numero de elementos no vetor!'
    
    elif type_ == 'root_n':
        try:         
            c_ =  np.sqrt(len(x_))
            if c_%1 != 0:
                c_ = int(c_) + 1

        except:
            return 'Problema com a raiz quadrada do numero de elementos no vetor!'

    elif type_ == 'inequality':
        try:         
            c_ = np.log(len(x_)) / np.log(2)
            if c_%1 != 0:
                c_ = int(c_) + 1

        except:
            return 'Problema com o logaritmo do número de elementos no vetor!'

    else:
        return 'Escolha uma opcao valida'
    
    min_ = np.min(x_)
    max_ = np.max(x_)
    at_ = max_ - min_

    if var_int:
        ac_ = int(at_/c_)
    else:
        ac_ = at_/c_

    lim_inf = min_
    lim_sup = lim_inf + ac_
    
    qtd_aux_01 = np.array([])
    limits_ = np.array([])

    for i in range(c_):
        if i == (c_ - 1):
            qtd = np.sum(np.logical_and(x_ >= lim_inf, x_ <= lim_sup))
            qtd_aux_01 = np.append(qtd_aux_01, int(qtd))
            class_ = '[' + str(np.round(lim_inf, 2)) + '; ' + str(np.round(lim_sup, 2)) + ']'
            class_ = class_.replace('.', ',')
            class_ = np.array(class_)
            limits_ = np.append(limits_, class_)

        else:
            qtd = np.sum(np.logical_and(x_ >= lim_inf, x_ < lim_sup))
            qtd_aux_01 = np.append(qtd_aux_01, int(qtd))
            class_ = '[' + str(np.round(lim_inf, 2)) + '; ' + str(np.round(lim_sup, 2)) + ')'
            class_ = class_.replace('.', ',')
            class_ = np.array(class_)
            limits_ = np.append(limits_, class_)

        lim_inf = lim_sup
        lim_sup = lim_inf + ac_

    df_obj = pd.DataFrame()
    df_obj['Int_Class'] = limits_
    
    qtd_aux_01_01 = [str(i).split('.')[0] for i in qtd_aux_01]
    df_obj['Abs_Freq'] = qtd_aux_01_01
    df_obj['Cum_Abs_Freq'] = np.cumsum(qtd_aux_01)
    
    df_obj['Relat_Freq'] = np.round(qtd_aux_01/len(x_) ,4)
    df_obj['Cum_Relat_Freq'] = np.round(np.cumsum(qtd_aux_01/len(x_)) ,4)    

    aux_percent_vet = [str(i)+'%' for i in np.round((qtd_aux_01/len(x_))*100 ,2)]
    df_obj['Percent_Freq'] = aux_percent_vet
    
    cum_aux_percent_vet = [str(i)+'%' for i in np.round(np.cumsum(qtd_aux_01/len(x_)*100),2)]
    df_obj['Cum_Percent_Freq'] = cum_aux_percent_vet
    
    dict_last_line = {'Int_Class': "Total",
                        'Abs_Freq': str(round(len(x_),0)),
                        'Cum_Abs_Freq': "-",
                        'Relat_Freq': str(1),
                        'Cum_Relat_Freq': "-",
                        'Percent_Freq': "100%",
                        'Cum_Percent_Freq': "-"}

    df_aux = pd.DataFrame(dict_last_line, index=[0])
    df_obj = pd.concat([df_obj, df_aux], axis = 0)
    df_obj = df_obj.reset_index().iloc[:, 1:]
    df_obj = df_obj.set_index('Int_Class')    


    if col_names == True:
        return df_obj[col_names_vet]

    else:
        if n_cols_ == 1:
            return pd.DataFrame(df_obj.iloc[:, 0:n_cols_])
        elif n_cols_ == 2:
            return pd.DataFrame(df_obj.iloc[:, 0:n_cols_])
        elif n_cols_ == 3:
            return pd.DataFrame(df_obj.iloc[:, 0:n_cols_])
        elif n_cols_ == 4:
            return pd.DataFrame(df_obj.iloc[:, 0:n_cols_])
        elif n_cols_ == 5:
            return pd.DataFrame(df_obj.iloc[:, 0:n_cols_])
        else:
            return pd.DataFrame(df_obj)
```

```python
### FUNÇÃO PARA CALCULAR ESTATÍSTICAS DESCRITIVAS
#import pandas as pd
#import numpy as np
#import math 

def desc_stats_fun(x__, casas__=2):
    len_aux = len(x__)
    if type(x__) == pd.Series:
        x__ = x__.values
        
    elif type(x__) != np.ndarray:
#        return 'Entre com uma variável do tipo np.array ou pd.serie.'
        return np.array([-9999, -9999, -9999, -9999, -9999, -9999, -9999, -9999, -9999, -9999,-9999, -9999, 0, len_aux])

    if pd.DataFrame(x__, columns=['x_'])['x_'].dtypes == 'O':
#        return 'Coloque um vetor (array) com valores numéricos.'
        return np.array([-9999, -9999, -9999, -9999, -9999, -9999, -9999, -9999, -9999, -9999,-9999, -9999, 0, len_aux])
    
    x__ = x__[~np.isnan(x__)]

    if len(x__)<1:
#        return 'Sem observações'    
        return np.array([-9999, -9999, -9999, -9999, -9999, -9999, -9999, -9999, -9999, -9999,-9999, -9999, 0, len_aux])

    def moment_cent(y__, t__):
        return np.mean((y__ - np.mean(y__))**(t__))
    
    min_ = np.min(x__) #mínimo
    max_ = np.max(x__) #máximo
    amp_total_ = (max_ - min_) #amplitude total
    mean_ = np.mean(x__) #média
    sd_ = np.std(x__) #desvio padrão
    dv_med_ = np.mean(np.abs(x__ - np.mean(x__))) #desvio médio
    var_ = np.var(x__) #variância
    median_ = np.median(x__) #mediana
    iiq_ = float(np.quantile(x__, 0.75) - np.quantile(x__, 0.25)) #intervalo interquartílico
    K_ = moment_cent(x__, 3)/(moment_cent(x__, 2)**(3/2)) # medida de assimetria
    C_ = moment_cent(x__, 4)/(moment_cent(x__, 2)**(2)) # medida de curtose
    CV_ = sd_/mean_
    n_obs = len(x__)
    return np.round(np.array([min_, max_, amp_total_, mean_, sd_, dv_med_, var_, median_, iiq_, K_, C_, CV_, n_obs, len_aux-n_obs]), casas__)
```

```python
### FUNÇÃO PARA CALCULAR ESTATÍSTICAS DESCRITIVAS PARA UMA LISTAS DE COLUNAS DE UM DATAFRAME
#import pandas as pd
#import numpy as np
#import math 

def desc_stats_df_fun(DF_, COL_LIST_):
    df_stats = pd.DataFrame()

    for i in COL_LIST_:
        df_loop_aux = pd.DataFrame(desc_stats_fun(DF_[i]), columns=[i])
        df_stats = pd.concat([df_stats, df_loop_aux], axis = 1)

    columns_stats = ['Min', 'Max', 'Amp_total', 'Mean', 'Sd', 'DV_med', 'Var', 'Median', 'Iiq', 'Assimetria', 'Curtose', 'CV', 'N_obs', 'N_na']
    df_stats = df_stats.T
    df_stats.columns = columns_stats

    return df_stats
```

```python
### Função para dist. freq. de uma variável qualitativa
#from collections import Counter
#import pandas as pd
#import numpy as np

def fun_ncat_aux(DF_, col_, cat_list_ = [], sort_bool_ = True, inf_bool_ = True):
#    DF_: DataFrame pandas com seus dados de interesse;
#    col_: nome da coluna que você quer trabalhar a distribuição de frequência;
#    cat_list_: Lista contendo o nome das categorias que você deseja que apareça na distribuição de frequência. Aqui você pode escolher a ordem que aparece as categorias; 
#    sort_bool_: Booleano para organizar as categorias por ordem alfanumérica ascendente. Aqui dispensamos a ordem colocada em cat_list_ se você escolher True;
#    inf_bool_: Booleano para inferir as categorias presentes na variável. Aqui dispensamos o parâmetro cat_list_ se você escolher True.


#    DF_NCAT_AUX = pd.DataFrame(dict(Counter(DF_[col_])), index=[0]).T.reset_index()
    DF_NCAT_AUX = pd.DataFrame(Counter(DF_[col_]), index=[0]).T.reset_index()
    DF_NCAT_AUX.columns = ['CATEGORIA', 'QTD']
    DF_NCAT_AUX['PROPORCAO'] = DF_NCAT_AUX['QTD']/DF_.shape[0]

    if inf_bool_:
        DF_NCAT_AUX_2 = pd.DataFrame({'CATEGORIA':list(set(DF_[col_]))})      

    else:
        DF_NCAT_AUX_2 = pd.DataFrame({'CATEGORIA':cat_list_})

    DF_NCAT_AUX_3 = DF_NCAT_AUX_2.merge(DF_NCAT_AUX, on='CATEGORIA', how='left').fillna(0)
    DF_NCAT_AUX_3['COLUNA'] = col_
    DF_NCAT_AUX_3 = DF_NCAT_AUX_3[['COLUNA', 'CATEGORIA', 'QTD', 'PROPORCAO']]
    DF_NCAT_AUX_3['PERCENTUAL'] = (DF_NCAT_AUX_3['PROPORCAO']*100).round(2)
    DF_NCAT_AUX_3['PROPORCAO'] = DF_NCAT_AUX_3['PROPORCAO'].round(4)

    if sort_bool_:
        return DF_NCAT_AUX_3.sort_values(by='CATEGORIA')

    else:
        return DF_NCAT_AUX_3 

#obs.: Faça um Loop junto com pd.concat e consiga um dataframe com a dist. freq de todas as suas variáveis categóricas.
```

```python
## ANÁLISE DE COMPONENTES PRINCIPAIS (PCA)
### VARIÂNCIA TOTAL
var_total = np.sum(DF_QUANT_FINAL[DF_QUANT_FINAL.columns[1:]].cov().values.diagonal())

### NÚMERO DE COMPONENTES
n_comp_aux = DF_QUANT_FINAL[DF_QUANT_FINAL.columns[1:]].shape[1]

### INTANCIANDO E AJUSTANDO A CLASSE/BOX DO PCA
pca_box = PCA(n_components = n_comp_aux, svd_solver = 'auto', random_state = 777, whiten = False)
pca_box_fitted = pca_box.fit(DF_QUANT_FINAL[DF_QUANT_FINAL.columns[1:]])
#obs: whiten = True é uma transformação a mais feita para garantir a independência dos componentes e pode melhorar a predição


### OBJETO AUXILIAR: seve para verificar variância total com os autovalores
bool_aux = (np.round(np.sum(pca_box_fitted.explained_variance_), 2) == np.round(var_total,2))
#obs: bool_aux é somente para mostrar que a soma dos autovalores tem que ser igual a variância total
print('Soma dos auto valores igual a variâncias total? {}! Isso indica que os autovalores estão corretos!'.format(bool_aux))
print('')

### VARIÂNCIA EXPLICADA PARA CADA COMPONENTE
DF_var_exp = pd.DataFrame(np.round(pca_box_fitted.explained_variance_ratio_, 4)*100).T
cp_name_list = ['CP_'+str(i+1) for i in range(n_comp_aux)]
DF_var_exp.columns = cp_name_list

### Dataframe com os valaores dos componentes para cada variável
DF_cp = pd.DataFrame(pca_box_fitted.components_).T
DF_cp.columns = cp_name_list

print('Total de variância explicada em cada componente')
DF_var_exp
```

##### **Passo 2:** Ajuste do modelo SArIMa para os dados de treino

```python
### DIVISÃO DA BASE EM TREINO E TESTE
DF_TRAIN = DF_TS_MES_ANO.iloc[:45,:]
DF_TEST = DF_TS_MES_ANO.iloc[45:,:]
```

```python
### MODELO SARIMA - QUANTIDADE DE CONTRATOS MENSAL
#obs: Configuração estimada: SARIMAX(1, 1, 0)x(0, 0, [], 12)

#### CONFIGURAÇÃO DO SARIMA
order_list = (1, 1, 0)  # Ordem do componente não sazonal
#seasonal_order_list = (0, 0, 0, 12)  # Ordem do componente sazonal

#### CONSTRUÇÃO DO MODELO
mod_sarima_vp = SARIMAX(endog = DF_TRAIN['soma_valor_principal']/m_aux_1, order=order_list).fit()

#### RESULTADOS DO MODELO
#print(mod_sarima_vp.summary())

### PREDIÇÃO DO MODELO PARA O ÚLTIMO TRIMESTRE
sarima_pred_vp = mod_sarima_vp.get_forecast(steps=3).predicted_mean
DF_SARIMA_PREDICT_IC_VP = mod_sarima_vp.get_forecast(steps=3).conf_int()
DF_SARIMA_PREDICT_IC_VP.columns = ['yhat_lower', 'yhat_upper']

### AVALIAÇÃO DO MODELO
DF_RESULT_SARIMA = medidas_ts(obs_ = (DF_TEST['soma_valor_principal']).values, pred_ = sarima_pred_vp*m_aux_1)
DF_RESULT_SARIMA['MODEL'] = 'Sarima'
DF_RESULT_SARIMA['MODEL_SIGLA'] = 'sarimax_package'
DF_RESULT_SARIMA['IC'] = True
```

##### **Passo 3:** Ajuste do modelo Prophet do Facebook

```python
### MODELO PROPHET
### DADOS PARA TREINO
DF_PROPHET_TRAIN_VP = DF_TRAIN[['ANO_MES', 'soma_valor_principal']]
DF_PROPHET_TRAIN_VP['ANO_MES'] = pd.to_datetime(DF_PROPHET_TRAIN_VP['ANO_MES'].apply(lambda val: val+'-01'))
DF_PROPHET_TRAIN_VP.columns = ["ds", "y"]#+col_X_aux
DF_PROPHET_TRAIN_VP['y'] = DF_PROPHET_TRAIN_VP['y']/m_aux_1

DF_PROPHET_TEST_VP = DF_TEST[['ANO_MES', 'soma_valor_principal']]
DF_PROPHET_TEST_VP['ANO_MES'] = pd.to_datetime(DF_PROPHET_TEST_VP['ANO_MES'].apply(lambda val: val+'-01'))
DF_PROPHET_TEST_VP.columns = ["ds", "y"]#+col_X_aux

### AJUSTE DO MODELO
model = Prophet()
#for regressor in col_X_aux:
#    model.add_regressor(regressor)
mod_prophet_VP = model.fit(DF_PROPHET_TRAIN_VP)

### PREDIÇÃO DO MODELO
DF_PROPHET_PREDICT_VP = mod_prophet_VP.predict(df=DF_PROPHET_TEST_VP[['ds']])
prophet_pred_VP = DF_PROPHET_PREDICT_VP.yhat.values
DF_PROPHET_PREDICT_IC_VP = DF_PROPHET_PREDICT_VP[['yhat_lower', 'yhat_upper']]

### AVALIAÇÃO DO MODELO
DF_RESULT_PROPHET = medidas_ts(obs_ =DF_PROPHET_TEST_VP['y'].values, pred_ = prophet_pred_VP*m_aux_1)
DF_RESULT_PROPHET['MODEL'] = 'Prophet'
DF_RESULT_PROPHET['MODEL_SIGLA'] = 'prophet'
DF_RESULT_PROPHET['IC'] = True
```

##### **Passo 4:** Ajuste dos modelos com Pycaret

```python
### BIBLIOTECAS
from pycaret import time_series as ts
from pycaret.time_series import TSForecastingExperiment

### OBJETO AUXILIAR PARA TRANSFORMAÇÃO DOS DADOS
m_aux_1 = 1000000

### DADOS
DF_AUTO_ML = DF_TRAIN[['soma_valor_principal']]/m_aux_1
DF_AUTO_ML = pd.concat([DF_AUTO_ML, DF_AUTO_ML.tail(1)], axis = 0).reset_index().iloc[:, 1:]

### DICIONÁRIO DE PARÂMETROS PARA O SETUP
dict_aux__nc = {
            'data':DF_AUTO_ML,
            'target': 'soma_valor_principal',
            'fold': 3,
            'fh': 1,
            'seasonal_period': 12,
#            'fold_strategy': 'expanding',
            'session_id': 123,
            'verbose': False, 
      }
#print('1')

### CAIXA COM A CLASSE
tsfe_box__nc = TSForecastingExperiment()
#print('2')

### AJUSTANDO O SETUP
tsfe_box__nc.setup(**dict_aux__nc)
print('Pycaret: Experimento TS inciado!')
print('')

### DATAFRAME PARA ARMAZENAR OS RESULTADOS
DF_RESULT_FINAL = pd.DataFrame()

### EXPERIMENTO - AJUSTE DOS MODELOS
for model_, name in tsfe_box__nc.models().reset_index()[['ID', 'Name']].values:
    ### AJUSTE DO MODELO
    model_ts_test = tsfe_box__nc.create_model(model_, verbose=False)

    ### PREDIÇÃO
    DF_PREDICT_PYCARET = tsfe_box__nc.predict_model(model_ts_test, return_pred_int=True, coverage=0.95, verbose=False, fh=3)*m_aux_1

    ### AVALIAÇÃO DOS MODELOS
    DF_RESULT = medidas_ts(obs_ = (DF_TEST['soma_valor_principal']).values, pred_ = DF_PREDICT_PYCARET['y_pred'].values)
    DF_RESULT['MODEL'] = name
    DF_RESULT['MODEL_SIGLA'] = model_
    if (np.sum(DF_PREDICT_PYCARET['lower'].isnull())>0) or (np.sum(DF_PREDICT_PYCARET['upper'].isnull())>0):
        DF_RESULT['IC'] = False
    else:
        DF_RESULT['IC'] = True

    DF_RESULT_FINAL = pd.concat([DF_RESULT_FINAL, DF_RESULT], axis = 0).reset_index().iloc[:, 1:]
    print(f'Modelo {name} avaliado.')

#DF_RESULT_FINAL
print('Pycaret: Experimento TS finalizado!')
```

Os modelos escolhidos são os 3 melhores usando Pycaret e que geram intervalo de confiança, e os outros 3 modelos que são referência de mercado.
